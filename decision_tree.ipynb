```python
input_data=movie.drop(columns=["Interest"])
input_data
```

```python
output_data=movie["Interest"]
output_data
```

```python
from sklearn.tree import DecisionTreeClassifier
movie_model=DecisionTreeClassifier()
```

```python
input_data_train,input_data_test,output_data_train,output_data_test=train_test_split(
    input_data,
    output_data,
    test_size=0.2,
    random_state=10
    
)
```

```python
movie_model=DecisionTreeClassifier(random_state=10)
movie_model.fit(input_data_train,output_data_train)
```

```python
movie_interest=movie_model.predict(input_data_test)
```

```python
comparison=pd.DataFrame({
    "Actual Interest":output_data_test,
    "Predicted Interest":movie_interest
})
comparison
```

```python
accuracy=accuracy_score(output_data_test,movie_interest)
accuracy
```
